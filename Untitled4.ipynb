```python
a = [[1, 2], [2, 3]]
b = [[5, 6], [6, 7]]
c = []

# Matrix Addition
for i in range(len(a)):
    row = []
    for j in range(len(a[i])):
        row.append(a[i][j] + b[i][j])
    c.append(row)

print(c)

for i in range(len(c)):
    for j in range(len(c[i])):
        print(c[i][j], end=" ")
    print()
```
